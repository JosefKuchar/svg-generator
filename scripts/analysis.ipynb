```python
from datasets import load_dataset

dataset = load_dataset("JosefKuchar/bezier-dataset", split="train")
```

```python
import json
from tqdm import tqdm
import matplotlib.pyplot as plt
curve_counts = []
for item in tqdm(dataset):
    shapes_data = json.loads(item["shapes"])
    curve_count = 0
    if len(shapes_data) == 0:
        continue
    for path in shapes_data[0]['paths']:
        curve_count += len(path['curves'])
    curve_counts.append(curve_count)
plt.figure(figsize=(10, 6))
plt.hist(curve_counts, bins=100, edgecolor='black')
plt.xlabel('Number of Curves')
plt.ylabel('Frequency (log scale)')
plt.title('Distribution of Curve Counts per Shape')
plt.yscale('log')
plt.grid(True, alpha=0.3)
plt.show()
```
